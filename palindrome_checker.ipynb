```python
def check_palindrome(input_str):
    stack = []
    queue = []
    
    # Step 1: Manually clean the string and convert to lowercase
    for char in input_str:
        # Check if character is a letter or number using ASCII ranges
        if ('a' <= char <= 'z') or ('A' <= char <= 'Z') or ('0' <= char <= '9'):
            
            # Convert uppercase letters to lowercase manually (ASCII offset is 32)
            if 'A' <= char <= 'Z':
                char = chr(ord(char) + 32)
                
            stack.append(char)  # Push to Stack (LIFO)
            queue.append(char)  # Enqueue to Queue (FIFO)
            
    # Step 2: Compare elements from Stack and Queue
    while len(stack) > 0:
        stack_char = stack.pop()      # Pops from the end (reverse order)
        queue_char = queue.pop(0)     # Pops from the front (original order)
        
        if stack_char != queue_char:
            return False # Mismatch found
            
    return True # Matched completely

if __name__ == "__main__":
    print("=== Simple Palindrome Checker ===")
    user_input = input("Enter a string or phrase: ")
    
    if check_palindrome(user_input):
        print("Result: It is a PALINDROME.")
    else:
        print("Result: It is NOT a palindrome.")
```
